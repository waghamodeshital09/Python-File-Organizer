```python
import os
import shutil

# Test folder location
FOLDER_PATH = os.path.join(
    os.path.expanduser("~"),
    "Desktop",
    "Organizer_Test"
)

# File categories
FILE_TYPES = {
    "Images": [".jpg", ".jpeg", ".png", ".gif", ".bmp", ".webp"],
    "Documents": [".pdf", ".docx", ".doc", ".txt", ".xlsx", ".pptx", ".md"],
    "Audio": [".mp3", ".wav", ".aac", ".flac"],
    "Videos": [".mp4", ".avi", ".mov", ".mkv"],
    "Archives": [".zip", ".rar", ".tar", ".gz"],
    "Scripts": [".py", ".js", ".sh", ".bat"]
}

# Create category folders
for folder in FILE_TYPES:
    os.makedirs(os.path.join(FOLDER_PATH, folder), exist_ok=True)

os.makedirs(os.path.join(FOLDER_PATH, "Others"), exist_ok=True)

# Count organized files
counts = {folder: 0 for folder in FILE_TYPES}
counts["Others"] = 0

# Organize files
for file in os.listdir(FOLDER_PATH):
    file_path = os.path.join(FOLDER_PATH, file)

    # Skip folders
    if not os.path.isfile(file_path):
        continue

    extension = os.path.splitext(file)[1].lower()
    destination_folder = "Others"

    for folder, extensions in FILE_TYPES.items():
        if extension in extensions:
            destination_folder = folder
            break

    destination = os.path.join(FOLDER_PATH, destination_folder, file)

    # Avoid overwriting a file with the same name
    if os.path.exists(destination):
        name, ext = os.path.splitext(file)
        number = 1

        while os.path.exists(
            os.path.join(FOLDER_PATH, destination_folder, f"{name}_{number}{ext}")
        ):
            number += 1

        destination = os.path.join(
            FOLDER_PATH, destination_folder, f"{name}_{number}{ext}"
        )

    try:
        shutil.move(file_path, destination)
        counts[destination_folder] += 1
        print(f"{file} -> {destination_folder}")
    except OSError as error:
        print(f"Could not move {file}: {error}")

# Final report
print("\n--- File Organizer Report ---")
for folder, count in counts.items():
    print(f"{folder}: {count} file(s)")

print("\nOrganization complete!")
```
